```python
#list in python
#list is an ordered and changeable collection that can store and list is a mutable
marks = [80,90,75,85]
print(marks)
```

```python
#accessing elements in a list
marks = [80,90,75,85]
print(marks[0])
print(marks[1])
print(marks[3])
```

```python
#change elements in a list
marks = [80,90,75]
marks[1] = 95
print(marks)
```

```python
#add elements to a list
marks = [80,90,75]
marks.append(85)
print(marks)
```

```python
#remove elements from a list
marks = [80,90,75]
marks.remove(90)
print(marks)
```

```python
#insert method
numbers = [10,20,30]
numbers.insert(1,15)
print(numbers)
```

```python
#extend method
a = [1,2,3]
b = [4,5,6]
a.extend(b)
print(a)
```

```python
#clear method
numbers = [10,20,30]
numbers.clear()
print(numbers)
```

```python
#index method
numbers = [10,20,30,40]
print(numbers.index(30))
```

```python
#count method
numbers = [10,20,20,30,20]
print(numbers.count(20))
```

```python
#sort method
numbers = [40,10,30,20]
numbers.sort()
print(numbers)
numbers.sort(reverse=True)
print(numbers)
```

```python
#reverse method
numbers = [10,20,30,40]
numbers.reverse()
print(numbers)
```

```python
#copy method
a = [1,2,3]
b = a.copy()
print(b)
```

```python
#dictionaries in python
#Dictionary is a collection  of key-values pairs that is unordered and mutable

student = {
    "name" : "Bhargavi" , 
    "age" :00 ,
    "course" : "Python"
    } 
print(student)
```

```python
#access elements in dic
print(student["name"])
print(student["age"])
print(student["course"])
```

```python
#change values in a dictionary
student["age"] = 11
print(student["age"])
```

```python
#add new data to a dictionary
student["city"] = "Vijayawada"
print(student)
```

```python
#remove data
#when we use pop method removes last inserted method in the list
student.pop("city")
print(student)
```

```python
#Dictionary
student = {
    "name": "Bhargavi",
    "age" : 21 ,
    "course" : "Python"
}

print(student.keys())
#keys() returns all the keys in the dictionary

print(student.values())
#values returns all the values in the dictionary

print(student.items())
#items() returns the values in the dictionary

print(student.get("name"))
#get() returns the value of the specified key

student.update({"age" : 22})
#update() updates the value of the specified key

print(student)

student.pop("age")
#pop() removes the specified key and its value

print(student)
```

```python
#popitem() removes the last inserted key-value pair 
student = {
    "name":"Bhargavi",
    "age" :21,
    "course" : "Python"
}
student.popitem()
print(student)

student = {
    "name" : "Bhargavi"
}
student.setdefault("age",21)
print(student)
```

```python
#clear method
student.clear()
print(student)
```

```python
#copy method
student = {
    "name" : "Bhargavi",
    "age" : 21
}
new_student = student.copy()
print(new_student)
```
