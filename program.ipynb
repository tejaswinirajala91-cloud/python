```python
#exchange values of two variables
a = int(input("Enter a:"))
b = int(input("Enter b:"))
  
temp = a
a=b
b = temp
print(a)
print(b)

a = a+b
b = a-b
a = a-b
print(a)
print(b)

a = a^b
b = a^b
a = a^b
print(a)
print(b)

a = a*b 
b = a/b
a = a/b
print(a)
print(b)

a,b = b,a
print(a)
print(b)
```
